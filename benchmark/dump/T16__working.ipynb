```python
import pandas as pd

# The attached file is available at /tmp/input/regions.xlsx
df = pd.read_excel('/tmp/input/regions.xlsx')

total_revenue = df['revenue'].sum()

# Group by region and find the sum of revenue for each region
revenue_by_region = df.groupby('region')['revenue'].sum()

# Find the region with the largest revenue
largest_region = revenue_by_region.idxmax()
largest_region_revenue = revenue_by_region.max()

print(f"Total Revenue: {total_revenue}")
print(f"Region with Largest Revenue: {largest_region}")
print(f"Revenue for Largest Region: {largest_region_revenue}")
```
