```python
import pandas as pd
import numpy as np

# 1. Generate 40 million order rows
num_rows = 40_000_000

# Generate random values between 10 and 500
order_values = np.random.uniform(10, 500, num_rows)

# Generate 12 region codes (e.g., 'Region_01' to 'Region_12')
region_codes = [f'Region_{i:02d}' for i in range(1, 13)]
regions = np.random.choice(region_codes, num_rows)

# Create a DataFrame
df = pd.DataFrame({
    'OrderValue': order_values,
    'Region': regions
})

# 2. Calculate the mean, 95th percentile, and total by region

# Overall Mean
overall_mean = df['OrderValue'].mean()

# Overall 95th percentile
percentile_95 = df['OrderValue'].quantile(0.95)

# Total by Region
total_by_region = df.groupby('Region')['OrderValue'].sum().reset_index()
total_by_region.columns = ['Region', 'TotalOrderValue']

# Prepare results for Excel
results_summary = pd.DataFrame({
    'Metric': ['Overall Mean Order Value', '95th Percentile Order Value'],
    'Value': [overall_mean, percentile_95]
})

# 3. Create an Excel workbook with these results
output_filename = '/tmp/output/capacity_model_simulation_results.xlsx'
with pd.ExcelWriter(output_filename, engine='xlsxwriter') as writer:
    results_summary.to_excel(writer, sheet_name='Summary Statistics', index=False)
    total_by_region.to_excel(writer, sheet_name='Total by Region', index=False)

print(f'Simulation results saved to {output_filename}')
```

```python
import pandas as pd
import numpy as np

# Set a seed for reproducibility
np.random.seed(42)

num_rows = 40_000_000
regions = [f'Region_{i+1}' for i in range(12)]

# Generate random order values between 10 and 500
order_values = np.random.uniform(10, 500, num_rows)

# Generate random region codes
region_codes = np.random.choice(regions, num_rows)

# Create a DataFrame
df = pd.DataFrame({'OrderValue': order_values, 'Region': region_codes})

# Calculate mean, 95th percentile for the entire dataset
mean_overall = df['OrderValue'].mean()
pct95_overall = df['OrderValue'].quantile(0.95)

# Calculate total by region
total_by_region = df.groupby('Region')['OrderValue'].sum().reset_index()
total_by_region.columns = ['Region', 'TotalOrderValue']

# Prepare output for Excel
results_summary = pd.DataFrame({
    'Metric': ['Overall Mean Order Value', 'Overall 95th Percentile Order Value'],
    'Value': [mean_overall, pct95_overall]
})

# Combine into a single Excel file with multiple sheets
output_filename = '/tmp/output/capacity_model_simulation_results.xlsx'
with pd.ExcelWriter(output_filename, engine='xlsxwriter') as writer:
    results_summary.to_excel(writer, sheet_name='Summary', index=False)
    total_by_region.to_excel(writer, sheet_name='Total by Region', index=False)

print(f'Simulation complete. Results saved to {output_filename}')
```
