```python
import pandas as pd

data = [
    ("Alpha", 1200, 45.00, 1050, 48.00),
    ("Beta", 800, 120.00, 1100, 118.00),
    ("Gamma", 2500, 18.00, 2300, 19.50)
]

df = pd.DataFrame(data, columns=['Product', 'Q1_units', 'Q1_price', 'Q2_units', 'Q2_price'])

df['Q1_revenue'] = df['Q1_units'] * df['Q1_price']
df['Q2_revenue'] = df['Q2_units'] * df['Q2_price']
df['Total_Change'] = df['Q2_revenue'] - df['Q1_revenue']

df['Volume_Effect'] = (df['Q2_units'] - df['Q1_units']) * df['Q1_price']
df['Price_Effect'] = (df['Q2_price'] - df['Q1_price']) * df['Q2_units']

df['Reconciliation_Check'] = df['Volume_Effect'] + df['Price_Effect']

# Ensure reconciliation
if not (df['Total_Change'] - df['Reconciliation_Check']).abs().sum() < 0.01:
    raise ValueError("Reconciliation failed! Total Change does not equal Volume Effect + Price Effect.")

output_df = df[['Product', 'Q1_revenue', 'Q2_revenue', 'Total_Change', 'Volume_Effect', 'Price_Effect']]

# Calculate totals
totals = output_df.sum(numeric_only=True)
totals['Product'] = 'Total'
output_df = pd.concat([output_df, pd.DataFrame([totals])], ignore_index=True)

output_filepath = '/tmp/output/q1_q2_revenue_bridge.xlsx'
writer = pd.ExcelWriter(output_filepath, engine='xlsxwriter')
output_df.to_excel(writer, sheet_name='Revenue_Bridge', index=False)

# Add a summary sheet
summary_data = {
    'Metric': [
        'Total Q1 Revenue',
        'Total Q2 Revenue',
        'Overall Total Change',
        'Overall Volume Effect',
        'Overall Price Effect'
    ],
    'Value': [
        output_df.loc[output_df['Product'] == 'Total', 'Q1_revenue'].iloc[0],
        output_df.loc[output_df['Product'] == 'Total', 'Q2_revenue'].iloc[0],
        output_df.loc[output_df['Product'] == 'Total', 'Total_Change'].iloc[0],
        output_df.loc[output_df['Product'] == 'Total', 'Volume_Effect'].iloc[0],
        output_df.loc[output_df['Product'] == 'Total', 'Price_Effect'].iloc[0]
    ]
}
summary_df = pd.DataFrame(summary_data)
summary_df.to_excel(writer, sheet_name='Summary', index=False)

writer.close()

print(f"Excel file created at {output_filepath}")

# Print key findings for the email summary
print(f"Overall Total Change: {output_df.loc[output_df['Product'] == 'Total', 'Total_Change'].iloc[0]:.2f}")
print(f"Overall Volume Effect: {output_df.loc[output_df['Product'] == 'Total', 'Volume_Effect'].iloc[0]:.2f}")
print(f"Overall Price Effect: {output_df.loc[output_df['Product'] == 'Total', 'Price_Effect'].iloc[0]:.2f}")
```
