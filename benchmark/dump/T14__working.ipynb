```python
import pandas as pd

data = {
    'Region': ['East', 'West', 'North', 'South'],
    'Revenue': [128400, 96750, 143200, 88900]
}
df = pd.DataFrame(data)

total_fy_revenue = df['Revenue'].sum()
largest_region_row = df.loc[df['Revenue'].idxmax()]
largest_region = largest_region_row['Region']

print(f'Total FY Revenue: {total_fy_revenue}')
print(f'Largest Region: {largest_region}')

# Prepare data for Excel output
output_df = df.copy()
summary_data = pd.DataFrame({
    'Metric': ['Total FY Revenue', 'Largest Region'],
    'Value': [total_fy_revenue, largest_region]
})

with pd.ExcelWriter('/tmp/output/regional_revenue_summary.xlsx', engine='xlsxwriter') as writer:
    output_df.to_excel(writer, sheet_name='Regional Data', index=False)
    summary_data.to_excel(writer, sheet_name='Summary', index=False)

print('Excel file created at /tmp/output/regional_revenue_summary.xlsx')
```
