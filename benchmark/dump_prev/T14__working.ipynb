```python
import pandas as pd
import io

data = {
    'Region': ['East', 'West', 'North', 'South'],
    'Revenue': [128400, 96750, 143200, 88900]
}
df = pd.DataFrame(data)

total_fy_revenue = df['Revenue'].sum()
largest_region_row = df.loc[df['Revenue'].idxmax()]
largest_region = largest_region_row['Region']
largest_region_revenue = largest_region_row['Revenue']

print(f"Calculated Total FY Revenue: {total_fy_revenue}")
print(f"Calculated Largest Region: {largest_region} with {largest_region_revenue} revenue")

# Create an Excel file with the data and summary
output = io.BytesIO()
writer = pd.ExcelWriter(output, engine='xlsxwriter')

# Write the detailed data
df.to_excel(writer, sheet_name='Regional Data', index=False)

# Write the summary on a separate sheet
summary_data = {
    'Metric': ['Total FY Revenue', 'Largest Region', 'Largest Region Revenue'],
    'Value': [total_fy_revenue, largest_region, largest_region_revenue]
}
summary_df = pd.DataFrame(summary_data)
summary_df.to_excel(writer, sheet_name='Summary', index=False)

writer.close()
output.seek(0)

with open('/tmp/output/regional_revenue_summary_corrected.xlsx', 'wb') as f:
    f.write(output.read())

print("Excel file created at /tmp/output/regional_revenue_summary_corrected.xlsx")
```
