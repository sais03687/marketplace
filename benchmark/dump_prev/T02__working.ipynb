```python
import pandas as pd

data = {
    'Product': ['Alpha', 'Beta', 'Gamma'],
    'Q1_units': [1200, 800, 2500],
    'Q1_price': [45.00, 120.00, 18.00],
    'Q2_units': [1050, 1100, 2300],
    'Q2_price': [48.00, 118.00, 19.50]
}
df = pd.DataFrame(data)

df['Q1_revenue'] = df['Q1_units'] * df['Q1_price']
df['Q2_revenue'] = df['Q2_units'] * df['Q2_price']
df['Total_change'] = df['Q2_revenue'] - df['Q1_revenue']

df['Volume_effect'] = (df['Q2_units'] - df['Q1_units']) * df['Q1_price']
df['Price_effect'] = (df['Q2_price'] - df['Q1_price']) * df['Q2_units']

# Calculate totals for summary
total_q1_revenue = df['Q1_revenue'].sum()
total_q2_revenue = df['Q2_revenue'].sum()
total_revenue_change = df['Total_change'].sum()
total_volume_effect = df['Volume_effect'].sum()
total_price_effect = df['Price_effect'].sum()

# Prepare for Excel output
output_df = df[['Product', 'Q1_units', 'Q1_price', 'Q1_revenue', 'Q2_units', 'Q2_price', 'Q2_revenue', 'Total_change', 'Volume_effect', 'Price_effect']]

# Add totals row
totals_row = pd.DataFrame([['Total', df['Q1_units'].sum(), '', total_q1_revenue, df['Q2_units'].sum(), '', total_q2_revenue, total_revenue_change, total_volume_effect, total_price_effect]], columns=output_df.columns)
output_df = pd.concat([output_df, totals_row], ignore_index=True)

output_file_path = '/tmp/output/Q1_Q2_Revenue_Bridge.xlsx'
output_df.to_excel(output_file_path, index=False)

print(f"Excel file saved to {output_file_path}")
print(f"Total Q1 Revenue: {total_q1_revenue:.2f}")
print(f"Total Q2 Revenue: {total_q2_revenue:.2f}")
print(f"Total Revenue Change: {total_revenue_change:.2f}")
print(f"Total Volume Effect: {total_volume_effect:.2f}")
print(f"Total Price Effect: {total_price_effect:.2f}")
```
