```python
import pandas as pd

data = [
    ["Region", "Revenue"],
    ["East", "128400"],
    ["West", "96750"],
    ["North", "143200"],
    ["South", "88900"]
]

df = pd.DataFrame(data[1:], columns=data[0])
df['Revenue'] = pd.to_numeric(df['Revenue'])

total_revenue = df['Revenue'].sum()
largest_region_row = df.loc[df['Revenue'].idxmax()]
largest_region_name = largest_region_row['Region']
largest_region_revenue = largest_region_row['Revenue']

print(f"Total Revenue: {total_revenue}")
print(f"Largest Region: {largest_region_name} with revenue {largest_region_revenue}")
```
