```python

import pandas as pd
import numpy as np

num_rows = 40_000_000

# Generate random order values and region codes without storing all in a single large DataFrame
# We will calculate aggregates as we go, or in a memory-efficient way.

# For overall mean and 95th percentile, we can sample or calculate on the fly if values are truly random.
# For total by region, we need to accumulate sums for each region.

# Initialize variables for aggregates
sum_values = 0
min_value = 10
max_value = 500
region_totals = {f'Region_{i+1}': 0 for i in range(12)}
all_values = [] # To calculate 95th percentile, we need all values or a large enough sample

# Generate and process in chunks to manage memory
chunk_size = 1_000_000

for i in range(0, num_rows, chunk_size):
    current_chunk_size = min(chunk_size, num_rows - i)
    
    # Generate random values for the current chunk
    values = np.random.randint(min_value, max_value + 1, current_chunk_size)
    regions = np.random.randint(1, 13, current_chunk_size) # 1 to 12
    
    sum_values += np.sum(values)
    all_values.extend(values.tolist()) # Extend list for percentile calculation
    
    # Accumulate region totals
    for region_idx in range(1, 13):
        region_key = f'Region_{region_idx}'
        region_filter = (regions == region_idx)
        region_totals[region_key] += np.sum(values[region_filter])

# Overall calculations
overall_mean = sum_values / num_rows
overall_95th_percentile = np.percentile(all_values, 95) # This might still be memory intensive

# Prepare data for Excel
results = {
    'Metric': ['Overall Mean', 'Overall 95th Percentile'] + list(region_totals.keys()),
    'Value': [overall_mean, overall_95th_percentile] + list(region_totals.values())
}
df_results = pd.DataFrame(results)

# Save to Excel
output_file_path = '/tmp/output/capacity_model_simulation_results.xlsx'
writer = pd.ExcelWriter(output_file_path, engine='xlsxwriter')
df_results.to_excel(writer, sheet_name='Simulation Results', index=False)
writer.close()

print(f"Simulation results saved to {output_file_path}")
```

```python
import pandas as pd
import numpy as np

num_rows = 40_000_000

# Generate random values between 10 and 500
values = np.random.uniform(10, 500, num_rows)

# Generate random region codes (0 to 11, then map to 'Region A', 'Region B', etc.)
region_codes_int = np.random.randint(0, 12, num_rows)
region_map = {i: f'Region {chr(65 + i)}' for i in range(12)}
regions = np.vectorize(region_map.get)(region_codes_int)

df = pd.DataFrame({'Value': values, 'Region': regions})

# Calculate mean, 95th percentile, and total by region
mean_by_region = df.groupby('Region')['Value'].mean()
pct_95_by_region = df.groupby('Region')['Value'].quantile(0.95)
total_by_region = df.groupby('Region')['Value'].sum()

# Calculate overall mean and 95th percentile
overall_mean = df['Value'].mean()
overall_pct_95 = df['Value'].quantile(0.95)

# Consolidate results
results_df = pd.DataFrame({
    'Mean Value': mean_by_region,
    '95th Percentile': pct_95_by_region,
    'Total Value': total_by_region
})

# Add overall statistics
overall_stats_df = pd.DataFrame({
    'Region': ['Overall'],
    'Mean Value': [overall_mean],
    '95th Percentile': [overall_pct_95],
    'Total Value': [df['Value'].sum()]
}).set_index('Region')

final_results = pd.concat([results_df, overall_stats_df])

# Prepare for Excel output
output_filename = '/tmp/output/capacity_model_simulation_results.xlsx'
writer = pd.ExcelWriter(output_filename, engine='openpyxl')
final_results.to_excel(writer, sheet_name='Simulation Results', index=True)
writer.close()

print(f"Results saved to {output_filename}")
```

```python
import pandas as pd
import numpy as np
import openpyxl

# Generate 40 million order rows
num_rows = 40_000_000

# Random values between 10 and 500
order_values = np.random.uniform(10, 500, num_rows)

# 12 region codes (Region 1 to Region 12)
region_codes = np.random.choice([f'Region {i}' for i in range(1, 13)], num_rows)

# Create DataFrame
df = pd.DataFrame({
    'OrderValue': order_values,
    'Region': region_codes
})

# Calculate mean, 95th percentile, and total by region
mean_by_region = df.groupby('Region')['OrderValue'].mean().reset_index()
mean_by_region.rename(columns={'OrderValue': 'Mean Order Value'}, inplace=True)

pct_95_by_region = df.groupby('Region')['OrderValue'].quantile(0.95).reset_index()
pct_95_by_region.rename(columns={'OrderValue': '95th Percentile Order Value'}, inplace=True)

total_by_region = df.groupby('Region')['OrderValue'].sum().reset_index()
total_by_region.rename(columns={'OrderValue': 'Total Order Value'}, inplace=True)

# Merge results
summary_df = mean_by_region.merge(pct_95_by_region, on='Region')
summary_df = summary_df.merge(total_by_region, on='Region')

# Ensure all regions are present, even if some had no orders (though highly unlikely with 40M rows)
# For robustness, let's explicitly include all 12 regions and fill NaNs if any.
all_regions = pd.DataFrame({'Region': [f'Region {i}' for i in range(1, 13)]})
summary_df = all_regions.merge(summary_df, on='Region', how='left').fillna(0) # Fill with 0 for missing values

# Write results to an Excel file
output_file_path = '/tmp/output/capacity_model_simulation_results.xlsx'
writer = pd.ExcelWriter(output_file_path, engine='openpyxl')
summary_df.to_excel(writer, sheet_name='Simulation Results', index=False)
writer.close()

print(f"Simulation results saved to {output_file_path}")

import os
file_id = os.listdir('/tmp/output')[0]
print(f"File ID: {file_id}")
```
