```python
import numpy as np
from scipy.stats import norm

# Given data
control_visitors = 8432
control_conversions = 421
variant_visitors = 8391
variant_conversions = 502

# Calculate conversion rates
cr_control = control_conversions / control_visitors
cr_variant = variant_conversions / variant_visitors

# Calculate relative lift
relative_lift = ((cr_variant - cr_control) / cr_control) * 100

# Perform two-sample proportion z-test for statistical significance
# Pooled proportion
p_pooled = (control_conversions + variant_conversions) / (control_visitors + variant_visitors)

# Standard error for pooled proportion
standard_error = np.sqrt(p_pooled * (1 - p_pooled) * (1/control_visitors + 1/variant_visitors))

# Z-statistic
z_stat = (cr_variant - cr_control) / standard_error

# P-value (two-tailed test)
p_value = 2 * (1 - norm.cdf(abs(z_stat)))

# Significance level
significance_level = 0.05

# Determine statistical significance
is_significant = p_value < significance_level

print(f"Control Conversion Rate: {cr_control:.4f}")
print(f"Variant Conversion Rate: {cr_variant:.4f}")
print(f"Relative Lift: {relative_lift:.2f}%")
print(f"Z-statistic: {z_stat:.4f}")
print(f"P-value: {p_value:.4f}")
print(f"Statistically Significant at {significance_level*100}% level: {is_significant}")

# Prepare data for Excel
import pandas as pd
df = pd.DataFrame({
    'Metric': [
        'Control Visitors',
        'Control Conversions',
        'Control Conversion Rate',
        'Variant Visitors',
        'Variant Conversions',
        'Variant Conversion Rate',
        'Relative Lift',
        'Z-statistic',
        'P-value',
        'Statistically Significant (alpha=0.05)'
    ],
    'Value': [
        control_visitors,
        control_conversions,
        f'{cr_control:.4f}',
        variant_visitors,
        variant_conversions,
        f'{cr_variant:.4f}',
        f'{relative_lift:.2f}%',
        f'{z_stat:.4f}',
        f'{p_value:.4f}',
        str(is_significant)
    ]
})

# Save to a temporary Excel file
df.to_excel('/tmp/output/checkout_test_results.xlsx', index=False)
```

```python
import pandas as pd
from statsmodels.stats.proportion import proportions_ztest
from io import BytesIO

# Data from the email
data = {
    'Variant': ['Control', 'Variant'],
    'Visitors': [8432, 8391],
    'Conversions': [421, 502]
}
df = pd.DataFrame(data)

# Calculate Conversion Rates
df['Conversion Rate'] = df['Conversions'] / df['Visitors']

control_cr = df[df['Variant'] == 'Control']['Conversion Rate'].iloc[0]
variant_cr = df[df['Variant'] == 'Variant']['Conversion Rate'].iloc[0]

# Calculate Relative Lift
relative_lift = ((variant_cr - control_cr) / control_cr) * 100

# Perform Z-test for two proportions
successes = df['Conversions'].tolist()
nobs = df['Visitors'].tolist()

# Ensure there are at least two samples for the z-test
if len(successes) == 2 and len(nobs) == 2:
    z_statistic, p_value = proportions_ztest(successes, nobs, alternative='two-sided')
else:
    z_statistic, p_value = None, None # Handle cases where data might be insufficient

alpha = 0.05
statistically_significant = p_value < alpha if p_value is not None else False

print(f"Control Conversion Rate: {control_cr:.4f}")
print(f"Variant Conversion Rate: {variant_cr:.4f}")
print(f"Relative Lift: {relative_lift:.2f}%")
print(f"Z-statistic: {z_statistic:.2f}")
print(f"P-value: {p_value:.4f}")
print(f"Statistically Significant (alpha={alpha}): {statistically_significant}")

# Create an Excel workbook in memory
output = BytesIO()
with pd.ExcelWriter(output, engine='xlsxwriter') as writer:
    # Write the summary results to a 'Summary' sheet
    summary_df = pd.DataFrame({
        'Metric': ['Control Conversion Rate', 'Variant Conversion Rate', 'Relative Lift', 'Z-statistic', 'P-value', f'Statistically Significant (alpha={alpha})'],
        'Value': [f'{control_cr:.4f}', f'{variant_cr:.4f}', f'{relative_lift:.2f}%', f'{z_statistic:.2f}', f'{p_value:.4f}', str(statistically_significant)]
    })
    summary_df.to_excel(writer, sheet_name='Summary', index=False)

    # Write the raw data to a 'Raw Data' sheet
    df.to_excel(writer, sheet_name='Raw Data', index=False)

# Save the in-memory Excel workbook to /tmp/output
with open('/tmp/output/checkout_test_results.xlsx', 'wb') as f:
    f.write(output.getvalue())

print("Excel workbook created at /tmp/output/checkout_test_results.xlsx")
```
