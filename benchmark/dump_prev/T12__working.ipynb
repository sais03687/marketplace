```python
import pandas as pd

data = {
    'Rep': ['Alice', 'Ben', 'Chidi'],
    'Revenue': [412000, 298000, 355000],
    'GrowthPct': [4, 41, 12],
    'MarginPct': [18, 22, 39]
}
df = pd.DataFrame(data)

# Define 'top performers' based on a weighted score or ranking
# For simplicity, let's rank each metric and sum the ranks.
# Higher rank is better (e.g., 1st place gets 3 points, 2nd gets 2, etc.)

df['RevenueRank'] = df['Revenue'].rank(ascending=False, method='min')
df['GrowthRank'] = df['GrowthPct'].rank(ascending=False, method='min')
df['MarginRank'] = df['MarginPct'].rank(ascending=False, method='min')

df['TotalScore'] = df['RevenueRank'] + df['GrowthRank'] + df['MarginRank']
df['OverallRank'] = df['TotalScore'].rank(ascending=True, method='min')

# Identify top performer(s) based on OverallRank
top_performer_reps = df[df['OverallRank'] == 1.0]['Rep'].tolist()

output_filename = '/tmp/output/rep_performance_analysis.xlsx'
writer = pd.ExcelWriter(output_filename, engine='xlsxwriter')
df.to_excel(writer, sheet_name='Performance Analysis', index=False)

# Add a summary sheet
summary_df = pd.DataFrame({
    'Metric': ['Overall Top Performer(s)'],
    'Value': [', '.join(top_performer_reps)]
})
summary_df.to_excel(writer, sheet_name='Summary', index=False)

writer.close()

print(f'Analysis saved to {output_filename}')
```
