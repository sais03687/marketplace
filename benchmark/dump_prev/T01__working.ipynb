```python
import pandas as pd

data = {
    'Region': ['North', 'South', 'East', 'West', 'Central'],
    'Visitors': [12480, 3200, 21750, 1540, 8030],
    'Conversions': [624, 288, 870, 200, 522]
}
df = pd.DataFrame(data)

df['Conversion Rate'] = (df['Conversions'] / df['Visitors'] * 100).round(2)

total_visitors = df['Visitors'].sum()
total_conversions = df['Conversions'].sum()
overall_conversion_rate = (total_conversions / total_visitors * 100).round(2)

overall_row = pd.DataFrame([['Overall', total_visitors, total_conversions, overall_conversion_rate]], columns=df.columns)
df_final = pd.concat([df, overall_row], ignore_index=True)

output_filename = '/tmp/output/conversion_rates_half.xlsx'
writer = pd.ExcelWriter(output_filename, engine='xlsxwriter')
df_final.to_excel(writer, sheet_name='Conversion Rates', index=False)
writer.close()

print(f"Overall Conversion Rate: {overall_conversion_rate}%")
print(f"File saved to {output_filename}")
```
